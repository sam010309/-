### 4.6. Запрос к нейросети с формулой в нотации TeX

**Запрос:** «Упрости выражение (x^2 − 1)/(x − 1) и выведи результат в нотации TeX».

**Ответ нейросети:**

$\frac{x^{2}-1}{x-1}=\frac{(x-1)(x+1)}{x-1}=x+1,\quad x\neq 1$

Исходный текст формулы: `$\frac{x^{2}-1}{x-1}=\frac{(x-1)(x+1)}{x-1}=x+1,\quad x\neq 1$`
