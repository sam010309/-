# Алгоритм проверки балла
Схема Mermaid и проверка граничных значений.

```python
import base64, zlib, json

mm = """flowchart TD
    A([Начало]) --> B[/Ввод балла b/]
    B --> C{b ≥ 60?}
    C -- Да --> D[/Вывод: зачёт/]
    C -- Нет --> E[/Вывод: незачёт/]
    D --> F([Конец])
    E --> F"""

# Mermaid Ink: UTF-8 -> urlsafe Base64
ink = "https://mermaid.ink/svg/" + base64.urlsafe_b64encode(mm.encode("utf8")).decode("ascii")

# Mermaid Live: JSON -> deflate -> Base64
c = zlib.compressobj(9, zlib.DEFLATED, -15)
raw = c.compress(json.dumps({"code": mm, "mermaid": {"theme": "default"}}).encode("utf8")) + c.flush()
live = "https://mermaid.live/edit#pako:" + base64.urlsafe_b64encode(raw).decode("ascii").rstrip("=")
print(ink); print(live)
```

## Проверка граничных значений
Для каждого входа сравниваем фактический ответ с ожидаемым.

```python
def result(b):
    if b >= 60:
        return "зачёт"
    return "незачёт"

for b, expected in [(0, "незачёт"), (59, "незачёт"),
                    (60, "зачёт"), (100, "зачёт")]:
    actual = result(b)
    assert actual == expected
    print(f"{b:3} баллов → {actual}")
```
